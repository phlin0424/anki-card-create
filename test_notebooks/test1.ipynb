````python
# ! This is an illustrative example, not executable code here.

import os
from google import genai
from PIL import Image

# 1. Set up the API Key and Client
# Assumes you have set the GEMINI_API_KEY environment variable
client = genai.Client()

# 2. Load the Image
# In a real script, you would replace 'path/to/your/image.jpg' with the actual path.
try:
    img = Image.open('/Users/lin-pei-hsuan/Documents/personal/anki-card-create/data/image.png') 
except FileNotFoundError:
    print("Error: Image file not found.")
    exit()
````

````python
from google import genai
from google.genai import types
# 3. Define the User Prompt
# This tells the model what to do. The System Instruction will handle the formatting.
user_prompt = "この画像を分析し、含まれる単語を抽出し、韓国語と日本語を並べてCSV形式で出力してください。"

# 4. Define the System Instruction (Crucial for format control)
# This forces the model to act as a pure data formatting machine.
system_instruction = (
    "あなたは、ユーザーの要求に応じて画像の内容を分析し、その結果を他のテキストを含まずに、純粋なCSVデータのみで返すAIです。出力をCSV形式（列の区切り文字はカンマ、行の区切り文字は改行）に厳密に従ってください。出力には、説明や追加のコメント、コードブロック（```csv...```）を含めないでください。"
)

# 5. Call the API with the System Instruction in config
response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=[user_prompt, img],
    config=types.GenerateContentConfig(
        system_instruction=system_instruction
    )
)

# 5. Process the Response
print("--- Image Context Extracted ---")
print(response.text)
````

````python
csv_data = response.text
````
